# Calendar Features

Calendar attributes are derived from the standardized `order_day` field.

The following features are created:

- Quarter
- Quarter Label
- Week Number
- Day of Month
- Day of Week
- Day Name
- Month Name
- Month Start Flag
- Month End Flag

These features provide reusable calendar dimensions for procurement analysis and downstream modeling.

```python
# ============================================================
# Calendar Feature Engineering
# ============================================================

feature_store["quarter"] = (
    feature_store["order_day"].dt.quarter
)

feature_store["quarter_label"] = (
    "Q" + feature_store["quarter"].astype(str)
)

feature_store["week_number"] = (
    feature_store["order_day"].dt.isocalendar().week.astype(int)
)

feature_store["day_of_month"] = (
    feature_store["order_day"].dt.day
)

feature_store["day_of_week"] = (
    feature_store["order_day"].dt.dayofweek
)

feature_store["day_name"] = (
    feature_store["order_day"].dt.day_name()
)

feature_store["month_name"] = (
    feature_store["order_day"].dt.month_name()
)

feature_store["is_month_start"] = (
    feature_store["order_day"].dt.is_month_start
)

feature_store["is_month_end"] = (
    feature_store["order_day"].dt.is_month_end
)

print("=" * 70)
print("CALENDAR FEATURES CREATED")
print("=" * 70)

calendar_features = [
    "quarter",
    "quarter_label",
    "week_number",
    "day_of_month",
    "day_of_week",
    "day_name",
    "month_name",
    "is_month_start",
    "is_month_end"
]

print("Features Created :", len(calendar_features))

display(
    feature_store[
        ["order_day"] + calendar_features
    ].head(10)
)
```

# Shipping Performance Features

Shipping performance is a critical procurement and operational KPI.

The existing dataset contains actual and scheduled shipping durations. These values are transformed into additional analytical indicators that quantify delivery variance and classify fulfillment performance.

The following features are created:

- `shipping_delay_days`
- `shipping_performance`
- `is_delayed`

A positive shipping delay indicates that actual shipping required more days than scheduled.

```python
# ============================================================
# Shipping Performance Features
# ============================================================

feature_store["shipping_delay_days"] = (
    feature_store["actual_shipping_days"]
    - feature_store["scheduled_shipping_days"]
)

feature_store["is_delayed"] = (
    feature_store["shipping_delay_days"] > 0
)

feature_store["shipping_performance"] = np.select(
    [
        feature_store["shipping_delay_days"] < 0,
        feature_store["shipping_delay_days"] == 0,
        feature_store["shipping_delay_days"] > 0
    ],
    [
        "Early",
        "On Time",
        "Delayed"
    ],
    default="Unknown"
)

print("=" * 70)
print("SHIPPING PERFORMANCE FEATURES")
print("=" * 70)

display(
    feature_store[
        [
            "actual_shipping_days",
            "scheduled_shipping_days",
            "shipping_delay_days",
            "is_delayed",
            "shipping_performance"
        ]
    ].head(10)
)
```

# Procurement Volume Features

Procurement quantity provides an important indicator of purchasing scale.

Instead of treating every transaction equally, transactions are categorized according to their purchased quantity.

The following features are created:

- `order_size_category`
- `is_bulk_order`

These features enable segmentation of procurement activity into small, medium, and large purchasing events.

The thresholds are derived from the distribution of transaction quantities rather than arbitrary business assumptions.

```python
# ============================================================
# Procurement Volume Features
# ============================================================

quantity_q1 = feature_store["quantity"].quantile(0.25)
quantity_q3 = feature_store["quantity"].quantile(0.75)

feature_store["order_size_category"] = np.select(
    [
        feature_store["quantity"] <= quantity_q1,
        feature_store["quantity"] <= quantity_q3
    ],
    [
        "Small",
        "Medium"
    ],
    default="Large"
)

feature_store["is_bulk_order"] = (
    feature_store["quantity"] > quantity_q3
)

print("=" * 70)
print("PROCUREMENT VOLUME FEATURES")
print("=" * 70)

print(f"Q1 Quantity : {quantity_q1}")
print(f"Q3 Quantity : {quantity_q3}")

display(
    feature_store["order_size_category"]
    .value_counts()
    .rename_axis("Order Size")
    .reset_index(name="Transactions")
)
```

### Observations

- The 25th percentile (Q1) of procurement quantity is **1 unit**, while the 75th percentile (Q3) is **3 units**.
- Based on these distribution-driven thresholds, **99,134 transactions (54.92%)** are classified as Small orders, **40,665 (22.53%)** as Medium orders, and **40,720 (22.56%)** as Large orders.
- The relatively high proportion of Small orders indicates that procurement activity is dominated by low-quantity transactions.
- Large orders represent approximately one-fifth of all transactions and can therefore be treated as an important segment for bulk procurement and demand analysis.
- The `is_bulk_order` indicator provides a simple binary flag for identifying transactions above the upper-quartile quantity threshold.

### Business Implications

- Small orders can be analyzed for opportunities to **consolidate purchases** and reduce transaction or procurement handling costs.
- Large orders should be monitored for **inventory capacity, supplier availability, fulfillment risk, and working-capital impact**.
- Comparing profitability, discounts, delivery performance, and supplier-related indicators across order-size segments can reveal whether larger procurement volumes generate operational or financial advantages.
- The `order_size_category` and `is_bulk_order` features can also be used as predictive-model inputs for **demand forecasting, delivery-risk prediction, and procurement optimization**.

### Validation

The engineered procurement volume features contain no missing values and are derived directly from the observed quantity distribution. The thresholds are therefore data-driven rather than based on arbitrary fixed business assumptions.

# Transaction Value Classification

Transaction value is an important indicator of procurement significance.

A business-oriented value classification is created using the distribution of `line_total`.

Transactions are categorized into:

- Low Value
- Medium Value
- High Value

The classification uses distribution-based thresholds to avoid imposing arbitrary monetary limits that may not generalize across datasets.

```python
# ============================================================
# Transaction Value Classification
# ============================================================

value_q1 = feature_store["line_total"].quantile(0.25)
value_q3 = feature_store["line_total"].quantile(0.75)

feature_store["transaction_value_category"] = np.select(
    [
        feature_store["line_total"] <= value_q1,
        feature_store["line_total"] <= value_q3
    ],
    [
        "Low Value",
        "Medium Value"
    ],
    default="High Value"
)

feature_store["is_high_value_transaction"] = (
    feature_store["line_total"] > value_q3
)

print("=" * 70)
print("TRANSACTION VALUE CLASSIFICATION")
print("=" * 70)

print(f"Q1 Line Total : {value_q1:.2f}")
print(f"Q3 Line Total : {value_q3:.2f}")

display(
    feature_store["transaction_value_category"]
    .value_counts()
    .rename_axis("Transaction Value")
    .reset_index(name="Transactions")
)
```

### Observations

- The 25th percentile (Q1) of `line_total` is **104.38**, while the 75th percentile (Q3) is **247.40**.
- Using these distribution-based thresholds, **45,306 transactions (25.10%)** are classified as Low Value, **90,134 (49.93%)** as Medium Value, and **45,079 (24.97%)** as High Value.
- Approximately half of all transactions fall into the Medium Value segment, indicating that the central portion of procurement activity is concentrated around typical transaction values.
- High-value transactions account for approximately one-quarter of the dataset and represent an important segment for financial and operational monitoring.
- The `is_high_value_transaction` flag provides a simple binary indicator for identifying transactions above the upper-quartile value threshold.

- The value classification and high-value flag can be used as predictive features for **procurement risk scoring, profitability analysis, and decision-support models**.

# Price Discount Features

The difference between the product list price and the actual unit selling price provides an independent measure of price reduction.

The following features are created:

- `price_discount_amount`
- `price_discount_pct`

The calculation uses the product list price and actual unit price already present in the Enterprise Dataset.

The resulting indicators provide a normalized view of pricing concessions across transactions.

```python
# ============================================================
# Price Discount Features
# ============================================================

feature_store["price_discount_amount"] = (
    feature_store["list_price"]
    - feature_store["unit_price"]
)

feature_store["price_discount_pct"] = np.where(
    feature_store["list_price"] > 0,
    (
        feature_store["price_discount_amount"]
        / feature_store["list_price"]
    ) * 100,
    0
)

print("=" * 70)
print("PRICE DISCOUNT FEATURES")
print("=" * 70)

display(
    feature_store[
        [
            "list_price",
            "unit_price",
            "price_discount_amount",
            "price_discount_pct"
        ]
    ].head(10)
)
```

# Daily Fuel Market Signals

Fuel prices are available as a complete daily time series in the Enterprise Dataset.

Because multiple procurement transactions may occur on the same day, a unique daily fuel series is created before calculating market movements.

The standardized `order_day` key is normalized to a calendar-day datetime on both the feature store and fuel lookup series to ensure consistent integration.

Daily fuel price movement is then calculated from the external daily series rather than from repeated procurement transactions.

Fuel prices are also classified into Low, Medium, and High levels using distribution-based thresholds.

```python
# ============================================================
# Prepare Daily Fuel Market Series
# ============================================================

fuel_daily = (
    enterprise[
        ["order_day", "fuel_price"]
    ]
    .copy()
)

# Standardize integration key
fuel_daily["order_day"] = pd.to_datetime(
    fuel_daily["order_day"],
    errors="coerce"
).dt.normalize()

# Create one record per calendar day
fuel_daily = (
    fuel_daily
    .drop_duplicates(subset=["order_day"])
    .sort_values("order_day")
    .copy()
)

# Calculate daily fuel price movement
fuel_daily["fuel_price_change_pct"] = (
    fuel_daily["fuel_price"]
    .pct_change()
    * 100
)

# Distribution-based thresholds
fuel_q1 = fuel_daily["fuel_price"].quantile(0.25)
fuel_q3 = fuel_daily["fuel_price"].quantile(0.75)

fuel_daily["fuel_price_category"] = np.select(
    [
        fuel_daily["fuel_price"] <= fuel_q1,
        fuel_daily["fuel_price"] <= fuel_q3
    ],
    [
        "Low",
        "Medium"
    ],
    default="High"
)

print("=" * 70)
print("DAILY FUEL MARKET SERIES")
print("=" * 70)

print("Unique Fuel Dates :", len(fuel_daily))
print("Date Range        :")
print(
    fuel_daily["order_day"].min(),
    "to",
    fuel_daily["order_day"].max()
)

print(f"\nFuel Q1 : {fuel_q1:.4f}")
print(f"Fuel Q3 : {fuel_q3:.4f}")

display(
    fuel_daily.head(10)
)
```

## Weather Intensity Features

Weather observations are transformed into operational indicators that can be

The engineering approach uses:

- Snowfall presence to identify snowy days.
- Distribution-based rainfall thresholds to classify rainfall intensity.
- Distribution-based wind-speed thresholds to classify wind intensity.

Distribution-based thresholds are used because the available weather variables

The resulting features provide a consistent analytical representation of
weather conditions without introducing arbitrary external thresholds.

```python
# ============================================================
# Weather Intensity Features
# ============================================================

rainfall_q1 = feature_store["rainfall"].quantile(0.25)
rainfall_q3 = feature_store["rainfall"].quantile(0.75)

wind_q1 = feature_store["max_wind_speed"].quantile(0.25)
wind_q3 = feature_store["max_wind_speed"].quantile(0.75)

feature_store["is_rainy_day"] = (
    feature_store["rainfall"] > 0
)

feature_store["is_snowy_day"] = (
    feature_store["snowfall"] > 0
)

feature_store["rainfall_intensity"] = np.select(
    [
        feature_store["rainfall"] <= rainfall_q1,
        feature_store["rainfall"] <= rainfall_q3
    ],
    [
        "Low",
        "Medium"
    ],
    default="High"
)

feature_store["wind_intensity"] = np.select(
    [
        feature_store["max_wind_speed"] <= wind_q1,
        feature_store["max_wind_speed"] <= wind_q3
    ],
    [
        "Low",
        "Medium"
    ],
    default="High"
)

feature_store["is_high_wind_day"] = (
    feature_store["max_wind_speed"] > wind_q3
)

weather_features = [
    "is_rainy_day",
    "is_snowy_day",
    "rainfall_intensity",
    "wind_intensity",
    "is_high_wind_day"
]

print("=" * 70)
print("WEATHER INTENSITY FEATURES")
print("=" * 70)

print(f"Rainfall Q1 : {rainfall_q1:.2f}")
print(f"Rainfall Q3 : {rainfall_q3:.2f}")
print(f"Wind Q1     : {wind_q1:.2f}")
print(f"Wind Q3     : {wind_q3:.2f}")

display(
    feature_store[
        [
            "order_day",
            "rainfall",
            "snowfall",
            "max_wind_speed"
        ] + weather_features
    ].head(10)
)
```

## Commodity Feature Duplicate Resolution Analysis

The feature naming validation identified duplicate commodity market signals
created during iterative feature-store enrichment.

The duplicated features appear with `_x` and `_y` suffixes as a result of
repeated merge operations.

Before removing or consolidating these columns, their values are compared
directly.

The objective is to determine whether:

1. `_x` and `_y` contain identical values and one copy can safely be removed.
2. The columns contain complementary values and require reconciliation.
3. The duplicated features originated from different calculations and require
   further investigation.

No columns are deleted until this comparison has been completed.

```python
# ============================================================
# Commodity Feature Duplicate Comparison
# ============================================================

commodity_base_features = [
    "energy_price_change_pct",
    "metals_price_change_pct",
    "agriculture_price_change_pct",
    "food_price_change_pct",
    "industrial_materials_price_change_pct"
]

print("=" * 70)
print("COMMODITY FEATURE DUPLICATE COMPARISON")
print("=" * 70)

comparison_results = []

for feature in commodity_base_features:

    col_x = f"{feature}_x"
    col_y = f"{feature}_y"

    x = feature_store[col_x]
    y = feature_store[col_y]

    comparison_results.append({

        "Feature": feature,

        "X Missing": x.isnull().sum(),

        "Y Missing": y.isnull().sum(),

        "Both Missing": (
            x.isnull() & y.isnull()
        ).sum(),

        "Values Equal": (
            x.eq(y) |
            (x.isnull() & y.isnull())
        ).all(),

        "Different Values": (
            ~(
                x.eq(y) |
                (x.isnull() & y.isnull())
            )
        ).sum()
    })

commodity_duplicate_comparison = pd.DataFrame(
    comparison_results
)

display(commodity_duplicate_comparison)
```

```python
# ============================================================
# Commodity Duplicate Value Comparison
# ============================================================

for feature in commodity_base_features:

    col_x = f"{feature}_x"
    col_y = f"{feature}_y"

    comparison_mask = ~(
        feature_store[col_x].eq(
            feature_store[col_y]
        )
        |
        (
            feature_store[col_x].isnull()
            &
            feature_store[col_y].isnull()
        )
    )

    print("=" * 70)
    print(feature)
    print("=" * 70)

    print(
        "Different Records :",
        comparison_mask.sum()
    )

    if comparison_mask.sum() > 0:

        display(
            feature_store.loc[
                comparison_mask,
                [
                    "year_month",
                    col_x,
                    col_y
                ]
            ].head(10)
        )
```
