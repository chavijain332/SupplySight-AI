## Executive Summary
This notebook focuses on acquiring daily foreign exchange rate data to assess the impact of currency fluctuations on international procurement. Exchange rates are treated as enterprise reference datasets and are later integrated with procurement data using standardized daily keys.

By incorporating exchange rate data, the analysis enables a more accurate evaluation of global procurement costs and currency exposure.

---

## Objective
- Acquire historical exchange rate data  
- Standardize daily exchange records  
- Prepare datasets for enterprise-level integration  

---

## Business Rationale
International procurement is highly sensitive to currency movements. Fluctuations in exchange rates directly impact:

- Import costs  
- Supplier pricing  
- International procurement expenditure  
- Currency risk and exposure  

Integrating daily exchange rates allows procurement transactions to be analyzed in the context of real-time currency variations, improving financial accuracy and decision-making.

---

## Data Sources
- **Provider:** Federal Reserve Economic Data (FRED)  
- **Datasets:**
  - USD → INR  
  - USD → EUR  
- **Frequency:** Daily  

---

## ETL Workflow
```
FRED
```

````python
import pandas as pd

from pandas_datareader import data as web
````

````python
start = "2015-01-01"
end = "2018-12-31"
````

````python
usd_inr = web.DataReader(
    "DEXINUS",
    "fred",
    start,
    end
)

usd_inr.head()
````


---

### 📊 Observations
- Dataset contains:
  - **Daily exchange rates**
  - Indexed by `DATE`
- Sample values:
  - 2015-01-02 → 63.27  
  - 2015-01-05 → 63.34  
- Missing values (`NaN`) observed on:
  - Weekends  
  - Public holidays (non-trading days)


### 📈 Insights
- Exchange rates provide **financial context** for:
  - Procurement costs  
  - Currency normalization  
- Daily granularity allows:
  - Time-series analysis  
  - Trend tracking  
- Missing values are **expected behavior** in financial datasets


### ⚠️ Risks / Considerations
- Presence of `NaN` values:
  - Must be handled before downstream usage
- Only USD → INR available:
  - May need expansion for multi-country analysis
- Data dependency on external API (FRED)

---



````python
usd_eur = web.DataReader(
    "DEXUSEU",
    "fred",
    start,
    end
)

usd_eur.head()
````

````python
print("="*50)
print("USD-INR")
print("="*50)

print("Start Date :", usd_inr["DATE"].min())
print("End Date   :", usd_inr["DATE"].max())

print()

print("="*50)
print("USD-EUR")
print("="*50)

print("Start Date :", usd_eur["DATE"].min())
print("End Date   :", usd_eur["DATE"].max())
````

````python
print("USD-INR Columns:")
print(usd_inr.columns)

print()

print("USD-EUR Columns:")
print(usd_eur.columns)
````

````python
datasets = {
    "USD-INR": usd_inr,
    "USD-EUR": usd_eur
}

for name, df in datasets.items():

    print("=" * 60)
    print(name)
    print("=" * 60)

    print(f"Rows       : {df.shape[0]}")
    print(f"Columns    : {df.shape[1]}")
    print(f"Duplicates : {df.duplicated().sum()}")

    print("\nMissing Values")
    print(df.isnull().sum())

    print("\nDate Range")
    print(df["DATE"].min(), "→", df["DATE"].max())

    print("\n")
````
