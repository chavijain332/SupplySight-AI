<h1 style="color: #1F4E78; font-weight: 700;">Step 4: Retrieve Public Holiday Data from the API</h1>

<h2 style="color: #1F4E78; font-weight: 700;">Objective</h2>
    
Fetch public holiday data for multiple countries and years using the Nager.Date Public Holidays API.

<h2 style="color: #1F4E78; font-weight: 700;">What this code does</h2>
                                                                     
- Iterates through each selected country.
- Retrieves holiday data for the years **2015–2018**.
- Sends an HTTP GET request to the API.
- Converts the returned JSON response into a Pandas DataFrame.
- Adds a new column indicating the country.
- Appends each year's data to the master DataFrame.
- Displays progress messages indicating successful retrieval or any errors encountered.

<h2 style="color: #1F4E78; font-weight: 700;">Why this step is important</h2>
    
Automating data retrieval ensures consistent collection of historical holiday information across multiple regions without manual downloads.

<h2 style="color: #1F4E78; font-weight: 700;">Output</h2>
    
A consolidated DataFrame containing public holiday records for all selected countries and years.

---



```python
holiday_df = pd.DataFrame()

for country in countries:

    for year in range(2015, 2019):

        url = f"https://date.nager.at/api/v3/PublicHolidays/{year}/{country}"

        response = requests.get(url)

        if response.status_code == 200:

            temp = pd.DataFrame(response.json())

            temp["Country"] = country

            holiday_df = pd.concat(
                [holiday_df, temp],
                ignore_index=True
            )

            print(f"✅ {country} {year}: {len(temp)} holidays")

        elif response.status_code == 204:

            print(f"⚠️ {country} {year}: No holiday data available")

        else:

            print(f"❌ {country} {year}: Error {response.status_code}")
```
